```python
#code 5
numbers = [14, 27, 9, 35, 18, 42]

target = int(input("Enter number to search: "))

found = False

for i in range(1, len(numbers)):
    if numbers[i] == target:
        found = True
        print("Found at index", i)

if found == True:
    print("Number found")
else:
    print("Number not found")
```
